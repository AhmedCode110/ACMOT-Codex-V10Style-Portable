# AC-MOT Baseline vs Final Video Demo
Run this notebook on a Google Colab T4. It creates a side-by-side VisDrone video with Baseline_Default on the left and Final AC-MOT on the right, including bounding boxes, track IDs, live demo ID-switch events, SCI/controller settings, and the official 17-sequence paper metrics in the footer.

The live `Demo IDS` counter is only a visualization aid linked to VisDrone ground truth by IoU >= 0.50. It is not the official TrackEval IDSW metric.


```python
from google.colab import drive
drive.mount('/content/drive', force_remount=False)

!pip install -q ultralytics==8.3.200 opencv-python-headless pandas numpy pyyaml

import os, shutil, subprocess, sys
REPO='/content/ACMOT-Codex-V10Style-Portable'
BRANCH='baseline-vs-acmot-video-demo'
if os.path.isdir(REPO):
    shutil.rmtree(REPO)
subprocess.run(['git','clone','-b',BRANCH,'https://github.com/AhmedCode110/ACMOT-Codex-V10Style-Portable.git',REPO], check=True)

DATASET='/content/drive/MyDrive/visdrone/VisDrone_Zips/VisDrone2019-MOT-test-dev/VisDrone2019-MOT-test-dev'
SEQUENCE='uav0000249_00001_v'  # dense scene; try uav0000306_00230_v too
OUTPUT_DIR='/content/drive/MyDrive/VisDrone_Results/ACMOT_VIDEO_DEMOS'
os.makedirs(OUTPUT_DIR, exist_ok=True)
OUTPUT=f'{OUTPUT_DIR}/{SEQUENCE}_Baseline_vs_Final_ACMOT.mp4'

cmd=[sys.executable, f'{REPO}/demo/acmot_video_demo.py',
     '--dataset', DATASET, '--sequence', SEQUENCE, '--output', OUTPUT,
     '--max-frames', '600', '--output-fps', '20', '--save-json']
print('RUNNING:', ' '.join(cmd))
subprocess.run(cmd, check=True)
print('\nREADY:', OUTPUT)
```
