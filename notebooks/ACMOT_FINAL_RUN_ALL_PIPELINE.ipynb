# AC-MOT Final Run-All Pipeline

This Colab-safe wrapper executes the full final pipeline in the current kernel: official TrackEval for the three replay finalists, automatic winner selection, T4 FP16 live benchmark, then official TrackEval on the final live predictions.


```python
# Load the validated pipeline cell sources from the original pipeline commit
import json, urllib.request, traceback

PIPELINE_URL = 'https://raw.githubusercontent.com/AhmedCode110/ACMOT-Codex-V10Style-Portable/02b50646d4e08ed0fdbc972a7130f3d135b7dd07/notebooks/ACMOT_FINAL_RUN_ALL_PIPELINE.ipynb'
print('Loading final AC-MOT pipeline...')
with urllib.request.urlopen(PIPELINE_URL) as r:
    pipeline = json.loads(r.read().decode('utf-8'))

code_cells = [c for c in pipeline.get('cells', []) if c.get('cell_type') == 'code']
print('Pipeline code cells:', len(code_cells))
if not code_cells:
    raise RuntimeError('No executable cells found in final pipeline source')

for i, cell in enumerate(code_cells, 1):
    src = ''.join(cell.get('source', []))
    print('\n' + '='*70)
    print(f'FINAL PIPELINE STEP {i}/{len(code_cells)}')
    print('='*70)
    try:
        exec(compile(src, f'<ACMOT final pipeline cell {i}>', 'exec'), globals(), globals())
    except Exception:
        print(f'FAILED AT FINAL PIPELINE STEP {i}')
        traceback.print_exc()
        raise

print('\n' + '='*70)
print('AC-MOT FINAL RUN-ALL PIPELINE COMPLETE')
print('='*70)
```
